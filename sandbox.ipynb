```python
from pathlib import Path
from wmdr.wmdr import WMDR10

mapping_file = "mappings/wmdr10_facility_to_acdd13.csv"

xml_files = [Path("tests/data/Blatten_20250729_0-20000-0-06725.xml"),
             Path("tests/data/20250504_0-20008-0-NRB.xml"),
]

result_path = Path("tests/results")

for xml_file in xml_files:
    wmdr10 = WMDR10(xml_file)
    wmdr10.export(base_path=result_path / xml_file.name.replace('.xml', '_simplified'))

    facility_acdd = wmdr10.facility_to_acdd(mapping_file)
    facility_acdd.export(path=result_path / xml_file.name.replace('.xml', '_facility_acdd.json'))
```
