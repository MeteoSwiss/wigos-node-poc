```python
from pathlib import Path
from wmdr.wmdr import WMDR10

mapping_file = "mappings/wmdr10_facility_to_acdd13.csv"

xml_files = [Path("tests/data/20250504_0-20008-0-NRB.xml"),
             Path("tests/data/Blatten_20250729_0-20000-0-06725.xml"),
             Path("tests/data/20200304_0-20000-0-06494.xml"),
             Path("tests/data/20250731_0-20000-0-06650.xml"),
]

result_path = Path("tests/results")

for xml_file in xml_files:
    wmdr10 = WMDR10(xml_file)
    wmdr10.export(path=result_path / xml_file.name.replace('.xml', ''))

    # convert to ACDD and export to .json, .geojson, or _oasis.json
    facility_acdd = wmdr10.facility_to_acdd(mapping_file)
    facility_acdd.export(path=result_path / xml_file.name.replace('.xml', '_facility_acdd.json'))
    facility_acdd.export(path=result_path / xml_file.name.replace('.xml', '_facility_acdd.geojson'), fmt='geojson')
    facility_acdd.export(path=result_path / xml_file.name.replace('.xml', '_facility_oasis.json'), fmt='oasis')

    # extract parts and store as possible input JSON
    # facility = wmdr10.extract(parts='facility')
    # observations = wmdr10.extract(parts='observations')
    # deployments = wmdr10.extract(parts='deployments')
    wmdr10.export(parts='header', path=result_path / xml_file.name.replace('.xml', '_header'))
    wmdr10.export(parts='facility', path=result_path / xml_file.name.replace('.xml', '_facility'))
    wmdr10.export(parts='observations', path=result_path / xml_file.name.replace('.xml', '_observations'))
    wmdr10.export(parts='deployments', path=result_path / xml_file.name.replace('.xml', '_deployments'))
    wmdr10.export(parts='observations', index=5, path=result_path / xml_file.name.replace('.xml', '_observation_5'))
    wmdr10.export(parts='deployments', index=3, path=result_path / xml_file.name.replace('.xml', '_deployment_3'))

    print(f"'{xml_file.name}' has been processed.")
```
